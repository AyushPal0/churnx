# ChurnX dataset exploration
This notebook uses the same validated source and cleaning function as training. Run from the repository root or open this notebook with the project virtual environment. Reproducible outputs are already generated by `python -m ml.train` in `reports/`.


```python
from pathlib import Path
import sys
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(root))
from ml.train import load_data
data = load_data(root / 'data/raw/Telco-Customer-Churn.csv')
data.describe(include='all')
```

```python
data['Churn'].value_counts(normalize=True).plot.bar(title='Historical churn distribution')
```

```python
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, feature in zip(axes, ['tenure', 'MonthlyCharges', 'TotalCharges']):
    data.boxplot(column=feature, by='Churn', ax=ax)
plt.suptitle('Numeric features against churn')
plt.tight_layout()
```

```python
from ml.train import eda
groups = eda(data, root / 'reports')
groups
```

## Interpretation
The target is moderately imbalanced, so accuracy alone is insufficient. This is a customer snapshot, not a monthly history. Categorical churn rates are associations. All encoding, scaling and imputation must be fitted inside training partitions; this notebook does not prepare a globally fitted training matrix.
